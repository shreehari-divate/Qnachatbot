Steps:
1] Get the API key from Groq and save it in .env

2] Access the API key 

3] Import the ChatGroq from which we can select the model and api  

4] Use system message and human message 

5] invoke the model om the message 

6] Use stroutputparser to fetch only the required message from the output

7] Chain the components model and parser and invoke the message on the chain

Same thing can be done with the help of prompt where a user can give the inputs

1-3 remains the same

4] Create a template for the system which should inform what we want. Here we can give a language as a variable 

5] CReate a chatprompt by taking system message and human message

6] Invoke this prompt by providing two things in dict, one is language variable defined in template

```python
import os
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()
# Access environment variables
OPEN_API_KEY = os.getenv("OPEN_API_KEY")
LANGCHAIN_API_KEY = os.getenv("LANGCHAIN_API_KEY")
LANGCHAIN_PROJECT = os.getenv("LANGCHAIN_PROJECT")

GROQ_API_KEY = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq


chat_groq = ChatGroq(model="llama-3.1-8b-instant", api_key=GROQ_API_KEY)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage

messages = [
    SystemMessage(content="Transalte the following question to Kannada."),
    HumanMessage(content="What is the capital of Karnataka?"),
]


result=chat_groq.invoke(messages)
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
parser.invoke(result)
```

```python
## Using LCEL to chain the components
chain = chat_groq | parser
chain.invoke(messages)
```

Using the prompt with lcel

```python
## Prompt templates with LCEL
from langchain_core.prompts import ChatPromptTemplate

generic_template = "Transalte the following question to {language} language"

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", generic_template),
        ("human", "{question}"),
    ]
)
```

```python
result=prompt.invoke({"language":"Kannada", "question":"What is the capital of Karnataka?"})
```

```python
chain = prompt|chat_groq|parser
chain.invoke({"language":"Kannada", "question":"What is the capital of Karnataka?"})
```
